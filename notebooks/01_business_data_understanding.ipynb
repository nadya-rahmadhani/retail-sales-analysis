#**1. Business Understanding**


###**1.1 Business Background**
Superstore merupakan perusahaan retail yang menjual berbagai kategori produk kepada pelanggan individu, perusahaan, maupun home office. Perusahaan memiliki banyak transaksi penjualan setiap hari sehingga menghasilkan data yang dapat dimanfaatkan untuk mengevaluasi performa bisnis. Agar perusahaan dapat mengambil keputusan yang lebih tepat, diperlukan analisis terhadap data penjualan untuk mengetahui performa penjualan, keuntungan, perilaku pelanggan, serta kategori produk yang memberikan kontribusi terbesar terhadap bisnis.

### **1.2 Business Problem**

Manajemen belum memiliki gambaran yang jelas mengenai performa penjualan dan profit di setiap
wilayah, kategori produk, serta segmen pelanggan. Tanpa analisis yang tepat, perusahaan akan
kesulitan menentukan strategi bisnis untuk meningkatkan pendapatan dan keuntungan.



### **1.3 Business Objectives**

- Menganalisis performa penjualan perusahaan.
- Mengidentifikasi kategori produk dengan performa terbaik.
- Mengetahui wilayah dengan penjualan dan profit tertinggi.
- Mengidentifikasi pelanggan dengan kontribusi terbesar.
- Memberikan rekomendasi bisnis berdasarkan hasil analisis.

### **1.4 Business Questions**

**Sales Performance**
1. Berapa total Sales yang dihasilkan perusahaan?
2. Berapa total Profit yang diperoleh perusahaan?
3. Bagaimana tren Sales dari waktu ke waktu?

**Product Analysis**

4. Kategori produk apa yang menghasilkan Sales terbesar?
5. Sub-category apa yang menghasilkan Profit terbesar?

**Customer Analysis**

6. Segmen pelanggan mana yang paling menguntungkan?
7. Siapa customer dengan kontribusi Sales terbesar?

**Regional Analysis**

8. Region mana yang menghasilkan Sales terbesar?
9. State mana yang memiliki Profit terendah?

**Discount Analysis**

10. Apakah Discount yang tinggi memengaruhi Profit?
